**1.¿Qué es Pandas?**

Pandas es una biblioteca de código abierto construida sobre el lenguaje de programación Python, diseñada específicamente para la manipulación y el análisis de datos. Proporciona estructuras de datos flexibles, rápidas y expresivas diseñadas para trabajar con datos estructurados (relacionales o etiquetados) de manera intuitiva. Sus estructuras principales son la Series (unidimensionales, similar a una columna) y el DataFrame (bidimensional, similar a una tabla de base de datos o una hoja de cálculo de Excel).

**2. ¿Para qué sirve?**

Pandas sirve para resolver tareas comunes en el procesamiento de datos dentro de proyectos de ingeniería, ciencia de datos y análisis financiero o administrativo. Sus funciones principales incluyen:

Limpieza y preparación de datos: Manejo de valores faltantes (NaN), duplicados, filtrado y transformación de tipos de datos.

Lectura y escritura de datos: Importación y exportación de información desde múltiples formatos como CSV, Excel, JSON, SQL y bases de datos.

Manipulación avanzada: Operaciones de agrupación (groupby), uniones (merge, join), tablas dinámicas (pivot_table) y ordenamiento.

Análisis estadístico básico: Cálculo rápido de medias, medianas, desviaciones estándar y correlaciones.

**3. ¿Cómo se utiliza / implementa?**

Instalación e importación: Se añade al entorno mediante el gestor de paquetes (pip install pandas) y se importa en el código utilizando el alias estándar import pandas as pd.

Uso de estructuras principales: Se implementa utilizando sus dos componentes fundamentales: las Series (arreglos unidimensionales o columnas) y los DataFrames (tablas bidimensionales con filas y columnas).

Carga de datos: Se alimenta de fuentes externas (como archivos CSV, Excel o bases de datos) mediante funciones integradas como pd.read_csv().

Procesamiento y análisis: Se ejecutan sus métodos optimizados para filtrar registros, agrupar datos (groupby), limpiar valores nulos y realizar cálculos estadísticos de forma directa sin necesidad de bucles complejos.

**4. Ejemplos de implementación aplicada**

Ejemplo 1: Crear una lista simple y ver los datos

```python
import pandas as pd

# 1. Crear un diccionario con datos
datos = {'Fruta': ['Manzana', 'Pera', 'Banano'], 'Cantidad': [10, 5, 20]}

# 2. Crear la tabla (DataFrame)
df = pd.DataFrame(datos)

# 3. Mostrar la tabla
print(df)
```

Ejemplo 2: Calcular el total de una columna

```python
import pandas as pd

datos = {'Producto': ['Camisa', 'Pantalón'], 'Precio': [20, 50]}

df = pd.DataFrame(datos)

# Sumar todos los precios
total = df['Precio'].sum()
print(f"Total a pagar: {total}")
```
